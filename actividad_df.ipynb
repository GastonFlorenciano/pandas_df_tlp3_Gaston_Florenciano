```python
import pandas as pd
import sqlite3

#Aqui agrega la conexión a la base de datos
conn = sqlite3.connect('Salaries.sqlite')

#Crea el dataframe con los datos del datasets
df = pd.read_sql_query("SELECT * FROM Salaries", conn)

conn.close()

df.head()
```

1. **Top 10 empleados con mayor salario total (incluyendo beneficios)**
Extrae e imprime un top 10 de expleados con mayor salario incluyendo los beneficios 


```python
#todo

df.sort_values('TotalPayBenefits', ascending=False, ignore_index=True)[['EmployeeName', 'TotalPayBenefits']].head(10) #Se usa "df" para seleccionar la tabla de la db y "sort_values" para ordenar los datos de la columna "TotalPayBenefits" de mayor a menor (ascending=False).Se muestran las columnas "EmployeeName" y "TotalPayBenefits" y se usa "head(10)" para mostrar solo los primeros 10 resultados.
```

2. **Filtrar empleados con más de 50,000 en horas extra, la columna a tratar es: (OvertimePay)**

```python
#todo

df["OvertimePay"] = pd.to_numeric(df["OvertimePay"], errors='coerce') #Convierte la columna "OvertimePay" a tipo numérico, manejando errores como NaN.

df[df["OvertimePay"] > 50000] #Filtra el dataframe para mostrar solo las filas donde "OvertimePay" es mayor a 50000.
```

3. **Contar cuántos empleados únicos hay por año**

```python
#todo

df.groupby("Year")["EmployeeName"].nunique() #Agrupa el dataframe por "Year" y cuenta el número de nombres únicos de empleados en cada año.
```

4. **Ver cuántos cargos únicos (JobTitle) existen y los 5 más comunes**

```python
#todo

cargos = df["JobTitle"].nunique() #Cuenta el número de títulos de trabajo únicos en el dataframe.

print('\nNúmero de títulos de trabajo únicos:\n') #Imprime el número de títulos de trabajo únicos.
print(cargos) #Imprime el número de títulos de trabajo únicos.

df["JobTitle"].value_counts().head() #Cuenta la frecuencia de cada título de trabajo y muestra los 5 más comunes.
```

5. **Mostrar el salario total (TotalPay) promedio por año**

```python
#todo

df.groupby("Year")["TotalPay"].mean().round(2) #Agrupa el dataframe por Año con el salario total promedio de cada uno.
```
